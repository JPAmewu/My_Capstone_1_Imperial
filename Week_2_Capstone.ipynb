```python

import numpy as np
import os

# Create folder to save Week 2 datasets
os.makedirs("week_2_data", exist_ok=True)
# Week 1 submitted query points
week_1_queries = {
    1: [0.375440, 0.950714],
    2: [0.375440, 0.950714],
    3: [0.444444, 0.666666, 0.333333],
    4: [0.555555, 0.444444, 0.222222, 0.111111],
    5: [0.224189, 0.846480, 0.879484, 0.878515],
    6: [0.728186, 0.154693, 0.732552, 0.693997, 0.564013],
    7: [0.045091, 0.528666, 0.329265, 0.105350, 0.434667, 0.641164],
    8: [0.273673, 0.260400, 0.073937, 0.078562, 0.862321, 0.230729, 0.106880, 0.352588]
}

# Week 1 returned outputs
week_1_outputs = {
    1: -1.560646704467778e-117,
    2: -0.03182956281754251,
    3: -0.04090761844901528,
    4: -8.727516493155957,
    5: 1088.8535114737463,
    6: -1.1520351120911565,
    7: 1.0510148516295004,
    8: 9.8157087929671
}

# Append Week 1 submitted query/output to starter data

for f in range(1, 9):

    X_old = np.load(f"initial_data_1/function_{f}/initial_inputs.npy")
    y_old = np.load(f"initial_data_1/function_{f}/initial_outputs.npy")

    X_new = np.array([week_1_queries[f]])
    y_new = np.array([week_1_outputs[f]])

    X_week2 = np.vstack([X_old, X_new])
    y_week2 = np.concatenate([y_old, y_new])

    np.save(f"week_2_data/function_{f}_inputs.npy", X_week2)
    np.save(f"week_2_data/function_{f}_outputs.npy", y_week2)

    print(f"\nFunction {f}")
    print("Old X shape:", X_old.shape)
    print("Old y shape:", y_old.shape)
    print("New Week 2 X shape:", X_week2.shape)
    print("New Week 2 y shape:", y_week2.shape)

```

```python
# Strategy used for each function in Week 1

import numpy as np

strategy = {
    1: "Bayesian Optimisation",
    2: "Bayesian Optimisation",
    3: "Bayesian Optimisation",
    4: "Bayesian Optimisation",
    5: "Manual Reasoning",
    6: "Bayesian Optimisation",
    7: "Bayesian Optimisation",
    8: "Bayesian Optimisation"
}

for i in range(1, 9):

    X = np.load(f"week_2_data/function_{i}_inputs.npy")
    y = np.load(f"week_2_data/function_{i}_outputs.npy")

    best_idx = np.argmax(y)

    print("\n" + "="*60)
    print(f"FUNCTION {i}")
    print(f"Strategy: {strategy[i]}")
    print("="*60)

    print("Number of Queries:", len(y))
    print("Best Output:", y[best_idx])
    print("Best Input Point:")
    print(X[best_idx])
```

```python
import numpy as np
import matplotlib.pyplot as plt

for i in range(1, 9):

    X = np.load(f"week_2_data/function_{i}_inputs.npy")
    y = np.load(f"week_2_data/function_{i}_outputs.npy")

    best_idx = np.argmax(y)

    query_numbers = np.arange(1, len(y) + 1)

    plt.figure(figsize=(8, 5))

    plt.plot(query_numbers, y, marker="o")
    plt.scatter(query_numbers[best_idx], y[best_idx], s=120)

    plt.title(f"Function {i}: Query Outputs")
    plt.xlabel("Query Number")
    plt.ylabel("Output Value")

    plt.grid(True)
    plt.show()

    print(f"Function {i}")
    print("Best Query Number:", best_idx + 1)
    print("Best Query Point:", X[best_idx])
    print("Best Output:", y[best_idx])
    print("-" * 60)
```

```python
import numpy as np
import matplotlib.pyplot as plt

for i in range(1, 9):

   X = np.load(f"week_2_data/function_{i}_inputs.npy")
   y = np.load(f"week_2_data/function_{i}_outputs.npy")

   n_features = X.shape[1]

   print(f"\nFUNCTION {i}")
   print(f"Dimensions: {n_features}")

   plt.figure(figsize=(5*n_features,4))

   for j in range(n_features):

        plt.subplot(1, n_features, j+1)

        plt.scatter(X[:,j], y)

        plt.xlabel(f"x{j+1}")
        plt.ylabel("Output (y)")
        plt.title(f"x{j+1} vs y")

        plt.grid(True)

        plt.tight_layout()
        plt.show()
```
